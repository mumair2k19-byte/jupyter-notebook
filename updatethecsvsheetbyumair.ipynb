```python
grading = data[1:]
```

```python
grading
```

```python
for row in grading:
    gra = float(row[5])

    if gra >= 80:
        grade = "A"
    elif gra >= 70:
        grade = "B"
    elif gra >= 60:
        grade = "C"
    elif gra >= 50:
        grade = "D"
    else:
        grade = "F"

    row.append(grade)
```
