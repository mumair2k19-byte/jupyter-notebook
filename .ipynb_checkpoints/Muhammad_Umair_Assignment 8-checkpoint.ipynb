#### Question 1
#### Employee Data Bank.



Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:

    def __init__(self, name, number, address, salary, post, joining_year, education, experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.joining_year = joining_year
        self.education = education
        self.experience = experience

    def describe_employee(self):
        print("Name:", self.name)
        print("Employee Number:", self.number)
        print("Address:", self.address)
        print("Salary:", self.salary)
        print("Post:", self.post)
        print("Year of Joining:", self.joining_year)
        print("Education:", self.education)
        print("Years of Experience:", self.experience)
        print("-----------------------------")

    
    def update_year_of_experience(self, years):
        self.experience += years
        print("Experience updated to:", self.experience, "years")

    # Method to increase salary
    def increase_salary(self, amount):
        self.salary += amount
        print("Salary increased to:", self.salary)




employee101 = Employee(
    "Ali",
    101,
    "Karachi",
    50000,
    "Software Developer",
    2022,
    "BS Computer Science",
    3
)

print(employee101.name)
print(employee101.number)
print(employee101.salary)
print(employee101.post)

employee101.describe_employee()

employee101.increase_salary(10000)

employee101.update_year_of_experience(1)



employee1 = Employee("Ahmed", 102, "Lahore", 60000, "Manager", 2020, "MBA", 5)

employee2 = Employee("Sara", 103, "Islamabad", 55000, "Accountant", 2021, "B.Com", 4)

employee3 = Employee("Asad", 104, "Karachi", 70000, "Developer", 2019, "BSCS", 6)

employee4 = Employee("Ayesha", 105, "Lahore", 45000, "Designer", 2023, "BS Design", 2)

employee5 = Employee("Bilal", 106, "Multan", 50000, "HR Officer", 2022, "BBA", 3)

employee6 = Employee("Hassan", 107, "Karachi", 80000, "Team Lead", 2018, "BSCS", 7)

employee7 = Employee("Zara", 108, "Islamabad", 65000, "Data Analyst", 2020, "BSIT", 5)

employee8 = Employee("Usman", 109, "Peshawar", 40000, "Support Officer", 2023, "BBA", 2)

employee9 = Employee("Fatima", 110, "Karachi", 75000, "Project Manager", 2019, "MBA", 6)

employee10 = Employee("Hamza", 111, "Lahore", 55000, "Web Developer", 2021, "BSCS", 4)



employee1.describe_employee()
employee2.describe_employee()
employee3.describe_employee()
employee4.describe_employee()
employee5.describe_employee()
employee6.describe_employee()
employee7.describe_employee()
employee8.describe_employee()
employee9.describe_employee()
employee10.describe_employee()



employee1.increase_salary(5000)
employee2.increase_salary(8000)
employee3.update_year_of_experience(2)
employee4.update_year_of_experience(1)
employee5.increase_salary(7000)
```
