```python
import pandas as pd
```

```python
df = pd.read_csv(r"cars.csv")
df
```

#### Check Duplicate Records

```python
duplicates = df[df.duplicated()]
print(duplicates)
```

#### Drop Duplicate Records

```python
drop = df.drop_duplicates()
print(drop)
```

#### Check missing or null values

```python
print(df.isnull().sum())
```

#### Drop missing or null values

```python
drop_missing = df.dropna()
print(drop_missing)
```

#### Show top 10 average MSRP (Price) per car brand (rounded figure)?

```python
top_10 = (
    df.groupby("Make")["MSRP"]
      .mean()
      .round(2)
      .sort_values(ascending=False)
      .head(10)
)

print(top_10)
```

#### Show the average MSRP (price) per years (rounded figure)?

```python
average_price = (df.groupby("Year")["MSRP"]
                   .mean()
                   .round(2)
                 
                
                
                
                )

print(average_price)
```

#### Show average Engine HP (power) rounded figure and MPG (efficiency) rounded figure per year?

Hint: Average of (city mpg + highway MPG)

```python
df["MPG"] = (df["city mpg"] + df["highway MPG"]) / 2

result = (
    df.groupby("Year")[["Engine HP", "MPG"]]
      .mean()
      .round(2)
)

print(result)
```

#### What percentage of cars are Compact, Midsize, or Large?

```python
percentage = df["Vehicle Size"].value_counts(normalize=True) * 100

print(percentage.round(2))
```

#### Which vehicle styles are the most sold (most common) in the dataset?

```python
most_common = df["Vehicle Style"].value_counts()

print(most_common)
```

#### Which car brands have the most vehicles priced at $40,000 or more and with a popularity score of at least 1000?

```python
result = (
    df[(df["MSRP"] >= 40000) & (df["Popularity"] >= 1000)]
    ["Make"]
    .value_counts()
)

print(result)
```
